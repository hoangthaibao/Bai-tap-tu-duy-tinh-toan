```python
def calculator(num1, num2, operat):
    if operat == '+':
        res = num1 + num2
    elif operat == '-':
        res = num1 - num2
    elif operat == '*':
        res = num1 * num2
    elif operat == '/':
        res = num1 / num2
    else:
        return "Invalid operator"

    return res 


a = float(input())
b = float(input())
op = input()

c = calculator(a, b, op)
print('%.2f' % c)
```

```python
def binary_search(a, l, r, k):
    res = -1
    while l <= r:
        mid = (l + r) // 2
        if a[mid] == k:
            res = mid
            r = mid - 1
        elif a[mid] < k:
            l = mid + 1
        else : r = mid - 1
    return res
a = list(map(int,input().split()))
k = int(input())
print(binary_search(a, 0, len(a) - 1, k))
            
```

```python
def nhi_phan(n):
    s = []
    while n != 0:
        s.append(n % 2)
        n //= 2
    s.reverse()
    return s
def check(c, d):
    a = nhi_phan(c)
    b = nhi_phan(d)
    while len(a) < len(b):
        a.insert(0, 0)
    while len(b) < len(a):
        b.insert(0, 0)
    sum = 0
    for i in range(len(a)):
        if a[i] != b[i]:
            sum += 1
    return sum
print(check(1, 4))
    
```
