```python
a = list(map(int, input().split()))

d = {'positives': 0, 'negatives': 0, 'zeros': 0}

for x in a:
    if x > 0:
        d['positives'] += 1
    elif x < 0:
        d['negatives'] += 1
    else:
        d['zeros'] += 1

print(d)
```

```python
d1 = {}
d2 = {}

# nhập dict 1
s1 = input().split()
for item in s1:
    k, v = item.split(":")
    d1[k] = int(v)

# nhập dict 2
s2 = input().split()
for item in s2:
    k, v = item.split(':')
    d2[k] = int(v)
d = d1.copy()
for i in d2:
    if i not in d:
        d[i] = d2[i]
    else : d[i] += d2[i]
for i in (sorted(d)):
    print(i, d[i])
    
```

```python
a = list(map(int, input().split()))
k = int(input())

res = []

n = len(a)
for i in range(n):
    for j in range(i+1, n):
        if a[i] + a[j] == k:
            res.append((a[i], a[j]))

# sắp xếp theo phần tử đầu tiên của tuple
res.sort(key=lambda x: x[0])

print(res)

```
