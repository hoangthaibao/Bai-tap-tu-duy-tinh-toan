```python
def binary_search(a, l, r, k):
    res = -1
    while l <= r:
        mid = (l + r) // 2
        if a[mid] == k:
            return mid
        elif a[mid] < k:
            l = mid + 1
        else:
            r = mid - 1
    return res
a = list(map(int,input().split()))
k = int(input())
print(binary_search(a, 0, len(a) - 1, k))
```

```python
def count_occurrences(arr, x):
    return arr.count(x)

arr = list(map(int, input().split()))
x = int(input())

print(count_occurrences(arr, x))
```

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(n - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

a = list(map(int, input().split()))
print(bubble_sort(a))
```

```python
def check(n):
    cnt = [0] * 10001
    for i in n:
        cnt[i] += 1
    max_val = -10 ** 18
    res = -1
    for i in n:
        if max_val < cnt[i]:
            max_val = cnt[i]
            res = i
    return max_val, res
n = list(map(int,input().split()))
print(check(n))
        
```

```python
def check(a, k):
    res = 0
    minn, maxx = min(a), max(a)
    max_val = -10 ** 18
    for i in range(minn, maxx + 1):
        if i in a:
            if max_val <= abs(k - i):
                max_val = abs(k - i)
                res = i
    return res
a = list(map(int,input().split()))
k = int(input())
print(check(a, k))
```
