```python
a = list(map(int,input().split()))
d = {}
for i in a:
    if i in d:
        d[i] += 1
    else : d[i] = 1
sum = 0
for x, y in d.items():
    if x < y:
        sum += y - x
    if x > y:
        sum += y
if sum == 0:
    print('YES')
else : print(f'NO {sum}')
```

```python
def swap(n):
    s = list(str(n))
    max_val = 10 ** 50
    for i in range(len(s)):
        for j in range(i + 1, len(s)):
            s[i], s[j] = s[j], s[i]
            max_val = min(max_val, int(''.join(s)))
            s[i], s[j] = s[j], s[i]
    return max_val
n = int(input())
print(swap(n))
```
