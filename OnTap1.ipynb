```python
num = list(map(int,input.split()))
k = int(input())
check = False
for i in range(len(num)):
    for j in range(i + 1, len(num)):
        if num[i] == num[j] and abs(j - i) <= k:
            check = True
            break
if check:
    print(True)
else:
    print(False)
            
```
