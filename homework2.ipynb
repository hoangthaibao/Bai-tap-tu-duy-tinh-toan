```python
a, b, m, c, d, n = map(int, input().split())

for i in range(0, min(m // a, n // c) + 1):
    for j in range(0, min(m // b, n // d) + 1):
        if a*i + b*j == m and c*i + d*j == n:
            print(i, j)
```
