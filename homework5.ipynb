```python
def binary_search(a, l, r, k):
    while l <= r:
        mid = (l + r) // 2
        if a[mid] == k:
            mid = k
            l = mid + 1
        elif a[mid] < k:
            l = mid + 1
        else : r = mid - 1
    return mid
a = list(map(int,input().split()))
k = int(input())
print(binary_search(a, 0, len(a) - 1, k))
```

```python
def binary_search(a, l, r, k):
    res = -1
    while l <= r:
        mid = (l + r) // 2
        if a[mid] == k:
            res = mid
            r = mid - 1
        elif a[mid] < k:
            l = mid + 1
        else : r = mid - 1
    if res == -1:
        return -1
    else : return res
    
    
a = list(map(int,input().split()))
k = int(input())
print(binary_search(a, 0, len(a) - 1, k))
```
